```python
%sql
DROP TABLE IF EXISTS sample.default.customer_feedback;

CREATE TABLE customer_feedback AS 
SELECT * FROM read_files(
  '/Volumes/sample/default/telecom/customer_feedback.csv',
  format => "csv",
  header => "true",
  inferSchema => "true"
);


```

```python
%sql
--Creating table.
DROP TABLE IF EXISTS sample.default.customer_payment;

CREATE TABLE sample.default.customer_payment (
  interaction_id STRING,
  customer_id STRING,
  interaction_type STRING,
  channel STRING,
  issue_category STRING,
  resolution_status STRING,
  interaction_date DATE,
  agent_id STRING
);
```
