```python
%sql
-- Create the table
DROP TABLE IF EXISTS sample.default.telecom_records;
    
CREATE TABLE IF NOT EXISTS sample.default.telecom_records
AS
SELECT * FROM read_files(
      '/Volumes/sample/default/telecom/',
      format => 'csv',
      inferSchema => true,
      header => true
);

-- Display the table
SELECT * FROM sample.default.telecom_records
```

```python
df = (
    spark.read.format("csv")
    .option("header", "true")
    .option("inferSchema", "true")
    .load(f'/Volumes/sample/default/telecom/device_management_dataset.csv')
)
df.show()

df.write.mode("overwrite").format("delta").saveAsTable(f"sample.default.device_detail")
```
